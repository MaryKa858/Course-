Среднее (μ) — оценка математического ожидания распределения. Характеризует центр распределения данных. В контексте выборки это наиболее вероятное (типичное) значение, вокруг которого группируются наблюдения. Служит прогнозом для нового значения из той же совокупности.

Стандартное отклонение (σ) — мера разброса данных относительно среднего. Показывает степень рассеивания значений. Маленькое σ означает плотную группировку данных вокруг μ; большое — значительный разброс.

Дисперсия (σ²) — квадрат стандартного отклонения. Количественная мера вариативности данных. Чем выше σ², тем больше неопределённость прогноза нового наблюдения.
